# Discrete Probability Distribution


```python
# importing core mathematical, statistical, and plotting libraries
import numpy as np # numerical calculation
import pandas as pd # data manipulation
import matplotlib.pyplot as plt # for plotting purposes
from scipy import stats # 

plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams["axes.edgecolor"] = "#008209"
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
np.random.seed(42)
```

## Bernaoulli Distribution 

{The probability with only two outcome yes or no.}

```python
p_bernoulli = 0.65 # 65% success of probabaility

k_bern = np.array([0,1]) # Support set k E (0,1)

# calculate theoritical PMF values p(x=k) via scipy.stats
pmf_bern = stats.bernoulli.pmf(k_bern,p_bernoulli)

# print mathematical properties

print("Success Parameter (p)",p_bernoulli)
print("p(x=0)",pmf_bern[0])
print("p(x=1)",pmf_bern[1])
print("")
```

## 2. Binomial Distribution
 - Binomial Distibution is what happens when you flip that coin multiple times and count how many total Heads you get.
 - "What is the probability of getting exactly $k$ successes out of $n$ independent tries?"

### the 4 

```python
from scipy.stats import binom

n = 5
p = 0.7

prob_4 = binom.pmf(k=4,n=n,p=p)
print("Exact probability of making 4 shots",prob_4)
```

## Poisson Distribution [Based on time]




```python
from scipy.stats import poisson

lambda_rate = 3
prob_2 = poisson.pmf(k=2,mu = lambda_rate)
print("Probability of getting exactly 2 orders :",prob_2,"or ",prob_2*100)
```
