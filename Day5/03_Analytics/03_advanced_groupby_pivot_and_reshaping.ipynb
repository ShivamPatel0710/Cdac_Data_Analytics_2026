```python
import pandas as pd
import numpy as np

df_agg = pd.DataFrame({
    'department' : ["IT","IT","HR","HR","Sales","Sales"],
    'salary' : [85000,92000,30000,54000,65000,70000],
    'tenure' : [3,5,2,6,4,7]
})
print(df_agg)
```

```python
print("Information :\n",df_agg.info())
print("description \n:",df_agg.describe())
```

```python
summary_df = df_agg.groupby('department').agg(
    mean_salary = ('salary','mean'),
    max_tenure = ('tenure','max')
).reset_index()
print(summary_df)
```

```python
# Pivot Table
df_pivot = pd.DataFrame({
    'device': ['Mobile','Desktop','Mobile','Desktop','Mobile','Desktop'],
    'region': ['North','North','South','South','North','South'],
    'sales' : [200,450,300,500,250,400]
})
print(df_pivot)
print("="*100)
pivot_res = df_pivot.pivot_table(index='device',columns="region",values='sales',aggfunc='sum',margins=True)
print(pivot_res)
```

```python
# melting dataset
df_wide = pd.DataFrame({
    'product' :['Widget A',"Widget B"],
    'Jan' : [100,150],
    'Feb' : [120,170]
})
print(df_wide)
```

```python
# melting pivoted table
melt_df = pd.melt(df_wide,id_vars='product',var_name='Month',value_vars=['Jan','Feb'],value_name='sales')
print(melt_df)
```

```python
# identify window function and rolling mean
np.random.seed(42)
dates = pd.date_range(start='2026-01-01',periods=10)
df_ts= pd.DataFrame({
    'date' : dates,
    'value' : np.random.randint(50,100,10)
})

df_ts['rolling_mean'] = df_ts['value'].rolling(window=3).mean()
print(df_ts)
```
