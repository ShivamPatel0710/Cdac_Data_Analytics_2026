```python
import pandas as pd
import numpy as np
from io import StringIO

csv_data = "id,val\n1,10\n2,20\n3,30\n4,10\n5,50"
chunk_size = 2
total_sum = 0

```

```python
for chunk in pd.read_csv(StringIO(csv_data),chunksize=chunk_size):
    print(chunk)
    
    total_sum += chunk['val'].sum()
print(f"Computed Total via Chunking {total_sum}")
```

```python
historical_state = pd.DataFrame({
    'day' : ['MON','TUE'],
    'Sales' : [100,150]
})
delta_batch = pd.DataFrame({
    'day' : ['WED'],
    'Sales' : [200]
})
# INCREMENTAL APPEND
update_state = pd.concat([historical_state,delta_batch],ignore_index=True)
print(update_state)
```
