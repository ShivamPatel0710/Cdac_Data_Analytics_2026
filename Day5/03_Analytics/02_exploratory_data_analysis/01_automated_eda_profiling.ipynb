### EDA (Exploratory Data Analysis)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_eda = pd.DataFrame({
    'age' : np.random.randint(18,70,200),
    'income' : np.random.normal(60000,15000,1200),
    'score' : np.random.rand(200)
})

print(df_eda.describe())
print("="*100)
print(df_eda.info())
print("="*100)
profile_summ = df_eda.describe().T['mean','std','min','max']
print(profile_summ)

plt.figure(figsize=(7,3))
sns.histplot(df_eda['income'],kde=True,color='dodgerblue')
```

```python
df_audit = pd.DataFrame({
    'id' : [1,2,2,4,5],
    'states' : ['active','pending','pending','closed',None],
    'amount' : [100,200,200,-50,500]
})

# Audit Marks
null_counts = df_audit.isna().sum()
duplicate_count = df_audit.duplicated().sum()
print(null_counts)
print(duplicate_count)
```
