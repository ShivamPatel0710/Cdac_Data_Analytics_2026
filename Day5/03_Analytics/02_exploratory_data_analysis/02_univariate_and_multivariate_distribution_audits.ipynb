```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
df_dist = pd.DataFrame({
    'department' : ['IT']*100 + ['HR'] * 100,
    'salary' : np.concatenate([np.random.normal(85000,10000,100),np.random.normal(55000,8000,100)])
})

print("Distributed Summary Statistics :",df_dist.groupby('department')['salary'].describe().T)

plt.figure(figsize=(7.5,3.5))
sns.violinplot(x='department',y = 'salary',hue = 'department',data=df_dist,palette='Set2',legend=False)
plt.title("Deep Dive Distribution Audit : Violin Plot Alanysis")
plt.ylabel("Salary")
plt.show()
```

```python
df_joint = pd.DataFrame({
    'feature_a' : np.random.normal(10,2,100),
    'feature_b' : np.random.normal(10,2,100),
    'category' : np.random.choice(["Group 1","Group 2"],100)

 })

print(df_joint[['feature_a','feature_b']].corr())

sns.pairplot(df_joint,hue='category',height=2.2,palette='husl')
plt.suptitle("Joint Distributrion Pairplot Analysis",y = 1.02 )
plt.show()
```

```python

df_breakdown = pd.DataFrame({
    'region' : ['North','North','South','South','East','East'],
    'sales' : [250,300,450,150,180,500]
})

segment_sum = df_breakdown.groupby('region')['sales'].agg(['median'])
print(segment_sum)
```
