```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
df_tax = pd.DataFrame({
    'age' : np.random.randint(20,60,100),
    'income' : np.random.normal(50000,15000,100)
})
df_tax.loc[df_tax['age'] < 30,'income'] = np.nan
print(df_tax)
```

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=5)
new = imputer.fit_transform(df_tax)
df = pd.DataFrame(new,columns=df_tax.columns)
```

```python
df[df['age']>30]
```

```python
df_imp=pd.DataFrame({
    'feature1':[1.0,2.0,np.nan,4.0,5.0],
    'feature2':[2.0,4.0,3.0,np.nan,10.0]
})
imputer=KNNImputer(n_neighbors=2)
pd.DataFrame(imputer.fit_transform(df_imp),columns=df_imp.columns)
```

```python
df_dup = pd.DataFrame({
    'customer_id': [101, 102, 101, 103],
    'name': ['Alice', 'Bob', 'Alice', 'Charlie'],
    'purchase_amt': [250, 150, 250, 300]
})

print(df_dup)
```

```python
df_mask = df_dup.duplicated()
df_dup["Duplicated"] = df_mask
print(df_dup)
```

```python
# removing all duplicate rows
df_clean = df_dup.drop_duplicates()
print(df_clean)
```
